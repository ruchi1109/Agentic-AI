```python
# Enhanced Job Application Crew - India Focused with Multiple Job Platforms
import warnings
warnings.filterwarnings('ignore')

from crewai import Agent, Task, Crew
import os
from dotenv import load_dotenv
from crewai_tools import (
    FileReadTool,
    ScrapeWebsiteTool,
    MDXSearchTool,
    SerperDevTool
)

load_dotenv()

openai_api_key = os.getenv('OPENAI_API_KEY')
os.environ["OPENAI_MODEL_NAME"] = 'gpt-3.5-turbo'
os.environ["SERPER_API_KEY"] = os.getenv('SERPER_API_KEY', '')

# Initialize tools
search_tool = SerperDevTool()
scrape_tool = ScrapeWebsiteTool()
read_resume = FileReadTool(file_path='Resume_Ruchi_Sharma.md')
semantic_search_resume = MDXSearchTool(mdx='Resume_Ruchi_Sharma.md')

# ===== ENHANCED INDIA-FOCUSED AGENTS =====

# 1. India Job Discovery Agent (Enhanced)
india_job_scraper = Agent(
    role="India Job Discovery Specialist",
    goal="Find and extract relevant job postings from Indian job platforms and international sites with India location filter",
    tools=[search_tool, scrape_tool],
    verbose=True,
    backstory=(
        "You are a master at navigating the Indian job market landscape. Your expertise "
        "includes major Indian job platforms like Naukri.com, Indeed India, LinkedIn India, "
        "Instahyre, Hirist, AngelList India, Glassdoor India, Monster India, TimesJobs, "
        "and company career pages with India locations. You understand the nuances of "
        "Indian job market, remote work opportunities, and can effectively filter for "
        "India-based positions across multiple platforms simultaneously."
    )
)

# 2. Job Links Aggregator Agent (NEW)
job_links_aggregator = Agent(
    role="Job Links Collection Specialist",
    goal="Collect and organize all job links from multiple sources and present them in an easily accessible format",
    tools=[search_tool, scrape_tool],
    verbose=True,
    backstory=(
        "You specialize in collecting and organizing job links from multiple sources. "
        "Your primary focus is to create comprehensive, well-organized lists of job "
        "opportunities with direct links, making it easy for candidates to quickly "
        "access and apply to multiple positions across different platforms."
    )
)

# 3. Market Intelligence Agent (India-focused)
india_market_researcher = Agent(
    role="India Tech Market Intelligence Analyst",
    goal="Research Indian tech market trends, salary ranges, and company insights for informed job applications",
    tools=[search_tool, scrape_tool],
    verbose=True,
    backstory=(
        "As an India market intelligence specialist, you have deep knowledge of the "
        "Indian tech industry landscape. You track salary trends in INR, startup "
        "ecosystem growth, funding rounds, tier-1/tier-2/tier-3 city opportunities, "
        "remote work trends, and industry shifts specific to the Indian market. "
        "You understand both service-based and product-based company landscapes in India."
    )
)

# 4. Company Deep Dive Agent (India-focused)
india_company_researcher = Agent(
    role="India Company Culture & Background Researcher",
    goal="Conduct thorough research on Indian companies and international companies with India presence",
    tools=[search_tool, scrape_tool],
    verbose=True,
    backstory=(
        "You specialize in uncovering the essence of companies operating in India - "
        "both Indian startups/enterprises and international companies with Indian "
        "offices. You research company culture, work-life balance, growth trajectories, "
        "recent funding rounds, expansion plans in India, remote work policies, "
        "and how they align with Indian work culture and candidate expectations."
    )
)

# Keep existing agents with minor modifications
profiler = Agent(
    role="Personal Profiler for Engineers",
    goal="Do incredible research on job applicants to help them stand out in the Indian job market",
    tools=[scrape_tool, search_tool, read_resume, semantic_search_resume],
    verbose=True,
    backstory=(
        "Equipped with analytical prowess, you dissect and synthesize information "
        "from diverse sources to craft comprehensive personal and professional profiles "
        "tailored for the Indian job market, considering local preferences and requirements."
    )
)

resume_strategist = Agent(
    role="Resume Strategist for Engineers",
    goal="Create resumes optimized for Indian job market and ATS systems used by Indian companies",
    tools=[scrape_tool, search_tool, read_resume, semantic_search_resume],
    verbose=True,
    backstory=(
        "With strategic understanding of Indian hiring practices and ATS systems, "
        "you excel at refining resumes to highlight relevant skills and experiences "
        "that resonate with Indian recruiters and international companies hiring in India."
    )
)

interview_preparer = Agent(
    role="Engineering Interview Preparer for Indian Market",
    goal="Create interview questions and talking points based on Indian company interview patterns",
    tools=[scrape_tool, search_tool, read_resume, semantic_search_resume],
    verbose=True,
    backstory=(
        "Your expertise covers both technical and cultural aspects of interviews in India. "
        "You understand the interview patterns of Indian service companies, product startups, "
        "and international companies with Indian operations, preparing candidates for diverse "
        "interview styles and expectations."
    )
)

# Keep other specialized agents with India context
skills_analyzer = Agent(
    role="Skills Gap Analysis Expert for Indian Market",
    goal="Identify skill gaps considering Indian job market demands and emerging tech trends",
    tools=[read_resume, semantic_search_resume, search_tool],
    verbose=True,
    backstory=(
        "You analyze skill gaps with deep understanding of the Indian tech landscape, "
        "including high-demand skills in Indian startups, service companies, and "
        "international firms. You provide actionable insights on skill development "
        "relevant to the Indian market context."
    )
)

cover_letter_writer = Agent(
    role="Cover Letter Specialist for Indian Companies",
    goal="Create compelling cover letters that resonate with Indian hiring managers and company culture",
    tools=[read_resume, semantic_search_resume, scrape_tool],
    verbose=True,
    backstory=(
        "You understand the communication style and preferences of Indian recruiters "
        "and hiring managers. Your cover letters balance professionalism with cultural "
        "sensitivity, demonstrating value propositions that resonate in the Indian context."
    )
)

application_strategist = Agent(
    role="India Application Strategy Coordinator",
    goal="Develop strategic approaches optimized for Indian job application processes and timelines",
    tools=[search_tool, scrape_tool],
    verbose=True,
    backstory=(
        "You understand the unique aspects of job hunting in India - festival seasons, "
        "financial year cycles, notice periods, and optimal application timing. Your "
        "strategies account for Indian business calendars and hiring patterns."
    )
)

network_analyzer = Agent(
    role="India Professional Network Analyst",
    goal="Identify networking opportunities within Indian tech communities and professional networks",
    tools=[search_tool, scrape_tool],
    verbose=True,
    backstory=(
        "You specialize in mapping Indian professional networks, including tech "
        "communities, alumni networks from Indian institutes, startup ecosystems, "
        "and professional groups that can provide valuable connections and insights."
    )
)

application_tracker = Agent(
    role="India Application Progress Manager",
    goal="Track applications considering Indian company response times and follow-up practices",
    tools=[search_tool],
    verbose=True,
    backstory=(
        "You understand Indian company hiring timelines, response patterns, and "
        "appropriate follow-up practices. Your tracking system accounts for longer "
        "response times and multi-stage interview processes common in Indian companies."
    )
)

# ===== ENHANCED TASKS FOR INDIA FOCUS =====

# Task 1: Comprehensive India Job Discovery
india_job_discovery_task = Task(
    description=(
        "Search for jobs across multiple Indian job platforms with India location filter. "
        "Target these platforms specifically:\n"
        "1. Naukri.com - search with location filter for India cities\n"
        "2. LinkedIn Jobs India - use 'India' location filter\n"
        "3. Indeed India - focus on Indian locations\n"
        "4. Instahyre - tech-focused Indian platform\n"
        "5. Hirist - IT/tech specific to India\n"
        "6. AngelList India - startup jobs\n"
        "7. Glassdoor India - with location filter\n"
        "8. Monster India - traditional job board\n"
        "9. TimesJobs - focus on Indian market\n"
        "10. Company career pages with India locations\n\n"
        "Search for these job titles: {target_job_titles}\n"
        "Apply India location filters: Major cities (Bangalore, Mumbai, Delhi, Hyderabad, Chennai, Pune, Kolkata) "
        "and also search for 'Remote India' and 'Work from Home India' positions.\n"
        "Do NOT limit to specific companies - search broadly across all companies hiring in India.\n"
        "Collect up to {max_jobs} relevant positions from diverse companies operating in India."
    ),
    expected_output=(
        "A comprehensive list of India-based job opportunities organized by platform, "
        "including direct job URLs, job titles, company names, locations in India, "
        "key requirements, salary ranges (if available), and application deadlines."
    ),
    agent=india_job_scraper,
    async_execution=True
)

# Task 2: Job Links Collection and Display
job_links_display_task = Task(
    description=(
        "From the comprehensive job discovery results, create a well-organized display "
        "of ALL job links categorized by platform. Present this as the first and most "
        "prominent output. Organize links by:\n"
        "1. Platform name (Naukri, LinkedIn, Indeed, etc.)\n"
        "2. Job title and company\n"
        "3. Location within India\n"
        "4. Direct application link\n"
        "5. Quick summary of key requirements\n\n"
        "Make this easily scannable with clear formatting and direct clickable links."
    ),
    expected_output=(
        "A prominently formatted, comprehensive list of all job links organized by platform "
        "with direct URLs, job details, and quick-scan information for immediate access and application."
    ),
    context=[india_job_discovery_task],
    agent=job_links_aggregator,
    output_file="india_job_links_directory.md"
)

# Task 3: India Market Intelligence
india_market_research_task = Task(
    description=(
        "Research current Indian tech market conditions for the target roles: {target_job_titles}. "
        "Focus on:\n"
        "- Salary ranges in INR for different cities (Tier-1, Tier-2 cities)\n"
        "- High-demand skills in Indian tech market\n"
        "- Growth trends in Indian startups vs established companies\n"
        "- Remote work adoption in India\n"
        "- Hiring patterns post-COVID in Indian tech sector\n"
        "- Popular tech stacks and frameworks in Indian companies"
    ),
    expected_output=(
        "An India-specific market intelligence report covering salary benchmarks in INR, "
        "trending skills in Indian market, company growth patterns, and current hiring "
        "conditions across different Indian cities and company types."
    ),
    agent=india_market_researcher,
    async_execution=True
)

# Task 4: India Company Research
india_company_research_task = Task(
    description=(
        "Research companies from the job discovery results, focusing on their India "
        "operations, work culture, growth trajectory, and employee reviews specifically "
        "for Indian offices. Include both Indian companies and international companies "
        "with significant India presence. Research work-life balance, career growth "
        "opportunities, and company culture in Indian context."
    ),
    expected_output=(
        "Detailed company profiles with India-specific insights including work culture, "
        "growth opportunities, salary benchmarks, employee satisfaction in Indian offices, "
        "and strategic priorities for each company's India operations."
    ),
    context=[india_job_discovery_task],
    agent=india_company_researcher,
    async_execution=True
)

# Task 5: Enhanced Profile Analysis for India Market
india_enhanced_profile_task = Task(
    description=(
        "Create a comprehensive candidate profile optimized for the Indian job market "
        "using GitHub URL ({github_url}), resume, and personal writeup ({personal_writeup}). "
        "Analyze how the candidate's skills and experience align with Indian market "
        "demands, including both service-based and product-based company requirements."
    ),
    expected_output=(
        "An India-market-optimized candidate profile highlighting skills and experiences "
        "most relevant to Indian tech companies, with positioning strategies for different "
        "types of Indian employers."
    ),
    context=[india_market_research_task],
    agent=profiler,
    async_execution=True
)

# Task 6: Skills Gap Analysis for India Market
india_skills_gap_task = Task(
    description=(
        "Analyze skill gaps between candidate profile and Indian job market requirements "
        "from discovered positions. Focus on skills highly valued in Indian tech companies, "
        "including both technical skills and soft skills relevant to Indian work culture. "
        "Provide learning resources and certification paths popular in India."
    ),
    expected_output=(
        "India-specific skills gap analysis with prioritized recommendations for skill "
        "development, popular Indian certifications, and learning resources accessible in India."
    ),
    context=[india_job_discovery_task, india_enhanced_profile_task],
    agent=skills_analyzer
)

# Task 7: India Job Requirements Analysis
india_job_analysis_task = Task(
    description=(
        "Analyze job requirements from Indian job postings, identifying common patterns "
        "in Indian tech companies vs international companies hiring in India. Categorize "
        "requirements by company type (service-based, product-based, startups, MNCs) "
        "and identify India-specific requirements or preferences."
    ),
    expected_output=(
        "Detailed analysis of job requirements categorized by Indian company types, "
        "with insights into different hiring patterns and preferences across the Indian tech landscape."
    ),
    context=[india_job_discovery_task],
    agent=profiler
)

# Task 8: India-Optimized Resume Creation
india_resume_task = Task(
    description=(
        "Create multiple resume versions optimized for different types of Indian employers:\n"
        "1. Service-based companies (TCS, Infosys, Wipro style)\n"
        "2. Product-based companies and startups\n"
        "3. International companies with India offices\n"
        "4. Indian unicorn startups\n"
        "Consider ATS systems commonly used in India and formatting preferences."
    ),
    expected_output=(
        "Multiple India-market-optimized resume versions tailored for different company "
        "types operating in India, with ATS-friendly formatting and relevant keyword optimization."
    ),
    output_file="india_optimized_resumes.md",
    context=[india_job_analysis_task, india_enhanced_profile_task, india_company_research_task],
    agent=resume_strategist
)

# Task 9: India Cover Letter Creation
india_cover_letter_task = Task(
    description=(
        "Create cover letter templates that resonate with Indian hiring managers, "
        "addressing cultural nuances and communication styles preferred in Indian "
        "professional contexts. Include templates for different company types and "
        "seniority levels commonly found in Indian market."
    ),
    expected_output=(
        "India-culturally-appropriate cover letter templates with examples that demonstrate "
        "understanding of Indian business communication styles and company expectations."
    ),
    output_file="india_cover_letters.md",
    context=[india_resume_task, india_company_research_task],
    agent=cover_letter_writer
)

# Task 10: India Application Strategy
india_application_strategy_task = Task(
    description=(
        "Develop application strategy considering Indian business cycles, festival seasons, "
        "financial year patterns, and typical notice periods. Include timing recommendations, "
        "prioritization based on Indian market dynamics, and follow-up strategies appropriate "
        "for Indian corporate culture."
    ),
    expected_output=(
        "Comprehensive India-specific application strategy with optimal timing, prioritization "
        "framework, and culturally appropriate follow-up approaches for maximum effectiveness."
    ),
    context=[india_job_discovery_task, india_company_research_task],
    agent=application_strategist
)

# Task 11: India Network Analysis
india_network_analysis_task = Task(
    description=(
        "Identify networking opportunities within Indian tech ecosystem including:\n"
        "- Indian Institute alumni networks\n"
        "- Local tech meetups and conferences\n"
        "- Indian startup ecosystem connections\n"
        "- Professional groups on LinkedIn India\n"
        "- Industry-specific Indian communities"
    ),
    expected_output=(
        "India-specific networking strategy with identified communities, events, and "
        "connection opportunities relevant to the Indian tech ecosystem."
    ),
    context=[india_company_research_task],
    agent=network_analyzer
)

# Task 12: India Interview Preparation
india_interview_task = Task(
    description=(
        "Create interview preparation materials specific to Indian company interview styles:\n"
        "- Technical rounds common in Indian companies\n"
        "- HR rounds and behavioral questions popular in India\n"
        "- Cultural fit questions for Indian work environment\n"
        "- Salary negotiation strategies for Indian market\n"
        "- Questions about notice periods, family background (common in India)"
    ),
    expected_output=(
        "Comprehensive India-specific interview guide covering technical, behavioral, "
        "and cultural aspects of interviews in Indian tech companies."
    ),
    output_file="india_interview_guide.md",
    context=[india_resume_task, india_company_research_task, india_job_analysis_task, india_skills_gap_task],
    agent=interview_preparer
)

# Task 13: India Application Tracking
india_tracking_task = Task(
    description=(
        "Create application tracking system accounting for longer response times common "
        "in Indian companies, multiple interview rounds, and appropriate follow-up timing "
        "considering Indian business practices and cultural norms."
    ),
    expected_output=(
        "India-optimized application tracking system with realistic timelines and "
        "culturally appropriate follow-up schedules for Indian job applications."
    ),
    output_file="india_application_tracker.md",
    context=[india_application_strategy_task, india_job_discovery_task],
    agent=application_tracker
)

# ===== INDIA-FOCUSED COMPREHENSIVE CREW =====
india_job_crew = Crew(
    agents=[
        # Primary agents
        india_job_scraper,
        job_links_aggregator,
        india_market_researcher,
        india_company_researcher,
        profiler,
        
        # Specialized agents
        skills_analyzer,
        resume_strategist,
        cover_letter_writer,
        interview_preparer,
        
        # Strategic agents
        application_strategist,
        network_analyzer,
        application_tracker
    ],
    tasks=[
        # Phase 1: Discovery and Links Display
        india_job_discovery_task,
        job_links_display_task,  # This will be the first prominent output
        
        # Phase 2: Research
        india_market_research_task,
        india_company_research_task,
        india_enhanced_profile_task,
        
        # Phase 3: Analysis
        india_skills_gap_task,
        india_job_analysis_task,
        
        # Phase 4: Content Creation
        india_resume_task,
        india_cover_letter_task,
        
        # Phase 5: Strategy and Preparation
        india_application_strategy_task,
        india_network_analysis_task,
        india_interview_task,
        india_tracking_task
    ],
    verbose=True
)

# ===== EXECUTION FUNCTION =====
def run_india_job_hunt(target_job_titles, github_url, personal_writeup, max_jobs=30):
    """
    Run the India-focused job hunting crew
    
    Args:
        target_job_titles: List of job titles to search for in India
        github_url: Candidate's GitHub profile URL
        personal_writeup: Personal description of the candidate
        max_jobs: Maximum number of jobs to analyze (default increased for broader search)
    
    Returns:
        Complete India job hunting package with all deliverables
    """
    
    job_hunt_inputs = {
        'target_job_titles': target_job_titles,
        'github_url': github_url,
        'personal_writeup': personal_writeup,
        'max_jobs': max_jobs
    }
    
    print(f"🇮🇳 Starting India-focused job hunt for {len(target_job_titles)} role types")
    print(f"🎯 Searching across all major Indian job platforms")
    print(f"📊 Analyzing up to {max_jobs} job opportunities across India")
    print(f"🌆 Covering major Indian tech hubs and remote opportunities")
    
    # Execute the crew
    result = india_job_crew.kickoff(inputs=job_hunt_inputs)
    
    print("✅ India job hunt analysis completed!")
    print("""
📄 Generated files:
- india_job_links_directory.md (🔝 ALL JOB LINKS ORGANIZED BY PLATFORM)
- india_optimized_resumes.md (Multiple resume versions for Indian market)
- india_cover_letters.md (India-appropriate cover letter templates)
- india_interview_guide.md (Interview prep for Indian companies)
- india_application_tracker.md (Tracking system with Indian timelines)
    """)
    
    return result

# ===== UPDATED EXAMPLE USAGE =====
if __name__ == "__main__":
    # Configuration for India job search
    india_job_titles = [
        "Machine Learning Engineer",
        "Data Scientist",
        "Software Engineer - Machine Learning", 
        "AI Engineer",
        "Senior Data Scientist",
        "Principal ML Engineer",
        "Staff Machine Learning Engineer",
        "MLOps Engineer",
        "Deep Learning Engineer",
        "Computer Vision Engineer",
        "NLP Engineer",
        "Data Engineer",
        "AI/ML Consultant"
    ]
    
    # No company restrictions - searching across all companies in India
    candidate_github = "https://github.com/ruchi1109"
    
    candidate_description = """Experienced Data Scientist/ML Engineer with 3+ years 
    building production-grade machine learning solutions, specializing in LLMs, GenAI, and data automation.
    Proven track record of delivering business-critical ML models that reduce processing time by 95% while
    maintaining high quality standards. Strong focus on practical, iterative development with expertise in 
    Python, R, cloud environments, and model deployment pipelines. Open to opportunities across India
    including remote work, and interested in both startup and established company environments.
    """
    
    # Run India-focused comprehensive analysis
    result = run_india_job_hunt(
        target_job_titles=india_job_titles,
        github_url=candidate_github,
        personal_writeup=candidate_description,
        max_jobs=30  # Increased for broader search across Indian market
    )
    
    # The job links will be prominently displayed at the top of outputs
    print("\n🔗 Check 'india_job_links_directory.md' for comprehensive list of all job links!")
```
